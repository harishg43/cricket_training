```python
df = pd.read_csv("Q11_cricket_training.csv")

print(df.head())
```

```python
print(df.info())
```

```python
print("Rows and Columns:", df.shape)
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()
df = df.drop_duplicates()

print("Cleaned dataset size:", df.shape)
```

```python
df["player_id"] = df["player_id"].astype(str).str.strip()
```

```python
print(df.groupby("training_program")["strike_rate"].describe())
```

```python
old = df[df["training_program"] == "Old"]["strike_rate"]
new = df[df["training_program"] == "New"]["strike_rate"]

t, p = ttest_ind(old, new)

print("T-value:", t)
print("P-value:", p)
```

```python
if p < 0.05:
    print("The new training programme significantly improves strike rate.")
else:
    print("The improvement is not statistically significant.")
```
